# DATA CLEANING

```python

print("Initial dataset info:")
print(df.info())
print("\nMissing values per column:")
print(df.isna().sum())

# Identify numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

for col in numeric_cols:
    mean_val = df[col].mean()
    df[col].fillna(mean_val, inplace=True)
    print(f"Filled missing values in '{col}' with mean = {mean_val:.3f}")

print("\nAfter cleaning:")
print(df.info())
print(df.head())

```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import SpectralClustering

# Step 1: Create a toy dataset (2 interleaving half-moons)
X, y_true = make_moons(n_samples=300, noise=0.05, random_state=42)

# Step 2: Apply Spectral Clustering
sc = SpectralClustering(n_clusters=2, affinity='nearest_neighbors', assign_labels='kmeans', random_state=42)
y_pred = sc.fit_predict(X)

# Step 3: Plot original vs clustered
plt.figure(figsize=(12,5))

# Original labels (ground truth)
plt.subplot(1,2,1)
plt.scatter(X[:,0], X[:,1], c=y_true, cmap='viridis')
plt.title("Original Data (True Classes)")

# Spectral clustering output
plt.subplot(1,2,2)
plt.scatter(X[:,0], X[:,1], c=y_pred, cmap='viridis')
plt.title("Spectral Clustering Result")

plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

# Select quantitative features
features = ['visits', 'age', 'income', 'children', 'exposure', 'health1', 'health2']
X = df[features].dropna()

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = model.fit_predict(X_scaled)
df.loc[X.index, 'cluster'] = labels

print(df[['rownames', 'cluster']].head())

# Plot dendrogram
linked = linkage(X_scaled, method='ward')
plt.figure(figsize=(12, 6))
dendrogram(linked,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=False)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()
```

## PCA

```python

# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 2: Load Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Medicaid1986.csv"
df = pd.read_csv(url)
print("Shape of dataset:", df.shape)
df.head()

# Step 3: Data Preprocessing
# Select only numeric columns for PCA
num_df = df.select_dtypes(include=[np.number]).dropna()
print("\n✅ Numeric DataFrame shape:", num_df.shape)

# Scale data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(num_df)

# Step 4: Apply PCA
pca = PCA()
pca_data = pca.fit_transform(scaled_data)

# Step 5: Explained Variance
explained_variance = pca.explained_variance_ratio_
cum_var = np.cumsum(explained_variance)

plt.figure(figsize=(8,5))
plt.bar(range(1, len(explained_variance)+1), explained_variance, alpha=0.6, label='Individual Component')
plt.plot(range(1, len(explained_variance)+1), cum_var, marker='o', color='red', label='Cumulative Variance')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.title('PCA Scree Plot - 22BDS0258')
plt.legend()
plt.grid(True)
plt.show()

# Step 6: 2D Visualization (Biplot for PC1 vs PC2)
pca_2 = PCA(n_components=2)
principal_components = pca_2.fit_transform(scaled_data)

pc_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
print("\n✅ First two principal components:")
pc_df.head()

# Biplot
plt.figure(figsize=(8,6))
sns.scatterplot(x='PC1', y='PC2', data=pc_df, s=50, alpha=0.7, color='blue')
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Biplot (PC1 vs PC2) - 22BDS0258")
plt.grid(True)
plt.show()

# Step 7: Display PCA Loadings (Optional)
loadings = pd.DataFrame(pca_2.components_.T,
                        columns=['PC1', 'PC2'],
                        index=num_df.columns)
print("\nPCA Loadings:")
print(loadings)
```
