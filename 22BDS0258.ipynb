```python
import pandas as pd

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Medicaid1986.csv"
df = pd.read_csv(url)

df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

# DIMENSIONS

```python
rows, cols = df.shape

print(f"Number of rows: {rows}")
print(f"Number of columns: {cols}")
```

# summary statistics

```python
print("Shape of dataset:", df.shape)
print("\nColumn names:\n", df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nFirst 5 rows:\n", df.head())
print("\nSummary Statistics:\n", df.describe(include="all"))
```

```python
df.describe()
```

# DATA HANDLING

```python
print("\nMissing Values:\n", df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

# DATA CLEANING

```python

print("Initial dataset info:")
print(df.info())
print("\nMissing values per column:")
print(df.isna().sum())

# Identify numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

for col in numeric_cols:
    mean_val = df[col].mean()
    df[col].fillna(mean_val, inplace=True)
    print(f"Filled missing values in '{col}' with mean = {mean_val:.3f}")

print("\nAfter cleaning:")
print(df.info())
print(df.head())

```

# UNIVARIATE ANALYSIS

```python
mean_income = df["income"].mean()
median_income = df["income"].median()
mode_income = df["income"].mode()

print("Mean income:", mean_income)
print("Median income:", median_income)
print("Mode income:", mode_income[0])
```

```python
#  First 30 largest categories (using 'ethnicity') & bar chart
df["ethnicity"].value_counts().nlargest(30).plot(kind='bar', figsize=(10,6))
plt.title("Number of observations by ethnicity")
plt.ylabel("Count")
plt.xlabel("Ethnicity")
plt.show()
```

```python
# Distribution plot for 'income'
sns.histplot(df["income"], kde=True)
plt.title("Distribution of Income")
plt.xlabel("Income")
plt.show()
```

```python
# Boxplot for 'income'
sns.boxplot(x="income", data=df)
plt.title("Boxplot of Income")
plt.show()
```

# BIVARIATE ANALYSIS

```python
# Relationship between 'income' and 'age'
plt.scatter(df["income"], df["age"])
plt.title("Scatter Plot: Age vs Income")
plt.xlabel("Income")
plt.ylabel("Age")
plt.show()

# Correlation (positive/negative relationships)
corr_income_age = np.corrcoef(df["income"], df["age"])
print("Correlation between income and age:", corr_income_age[0,1])

corr_income_children = np.corrcoef(df["income"], df["children"])
print("Correlation between income and children:", corr_income_children[0,1])
```

```python
#  Box plot: married status vs income
sns.boxplot(x="married", y="income", data=df)
plt.title("Income Distribution by Married Status")
plt.show()
```

```python
#  Box plot: gender vs income
sns.boxplot(x="gender", y="income", data=df)
plt.title("Income Distribution by Gender")
plt.show()
```

# MULTIVARIATE ANALYSIS

```python
# Pair plot with regression lines
sns.pairplot(df, vars=['age', 'income', 'children'], kind="reg")
plt.suptitle("Pair Plot with Regression", y=1.02)
plt.show()
```

```python
# Pair plot with hue (categorical coloring)
sns.set(style="ticks", color_codes=True)
sns.pairplot(df, height=2, vars=['age', 'income', 'children'], hue="married")
plt.suptitle("Pair Plot with Hue", y=1.02)
plt.show()
```

```python
# Pearson correlation between two variables
from scipy import stats
corr, p_value = stats.pearsonr(df["income"], df["age"])
print(f"Correlation: {corr:.4f}")
print(f"P-value: {p_value:.4e}")
```

```python
# Correlation matrix for all numeric columns
correlation = df.corr(method='pearson', numeric_only=True)
print(correlation)
```

```python
# Heatmap of correlation matrix
plt.figure(figsize=(10,8))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f",
            xticklabels=correlation.columns,
            yticklabels=correlation.columns)
plt.title("Correlation Heatmap - Medicaid 1986 Dataset")
plt.show()
```
